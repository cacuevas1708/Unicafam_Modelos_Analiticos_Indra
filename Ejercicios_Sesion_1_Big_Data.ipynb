```python
# Ejercicio 3: Primer análisis de ventas
```

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
n = 200
```

```python
ventas = pd.DataFrame({
    "fecha": pd.date_range("2026-01-01", periods=n, freq="D"),
    "ciudad": np.random.choice(["Bogotá", "Medellín", "Cali", "Barranquilla"], n),
    "canal": np.random.choice(["Tienda", "Web", "App"], n),
    "unidades": np.random.randint(1, 20, n),
    "precio": np.random.choice([25000, 48000, 99000], n),
})
```

```python
ventas["total"] = ventas["unidades"] * ventas["precio"]
```

```python
ventas.head()
```

```python
# ¿Cuál es el total vendido por canal (Tienda, Web, App)?

ventas.groupby("canal")["total"].sum()
```

```python
# ¿Cuál es el ticket promedio por ciudad?

ventas.groupby("ciudad")["total"].mean()
```

```python
# Días con mas ventas

ventas["dia_semana"] = ventas["fecha"].dt.day_name()

ventas.groupby("dia_semana")["total"].sum().idxmax()
```

El canal Web lidera las ventas, Bogotá registra la mayor facturación y Medellín presenta el ticket promedio más alto.
