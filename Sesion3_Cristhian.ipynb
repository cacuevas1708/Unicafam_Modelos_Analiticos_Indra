```python
# Ejercicio 9: Dividir la cuenta

cuenta = 600000
propina = 10
amigos = 3
```

```python
# Crear función para que sume la propina

def con_propina(cuenta, porcentaje):

    total = cuenta + (cuenta * porcentaje / 100)

    return total
```

```python
# Crear función para dividir la cuenta

def por_persona(total, personas):

    return total / personas
```

```python
# Calcular el total con propina

total = con_propina(cuenta, propina)

print("Total con propina:", total)
```

```python
# Calcular cuanto paga cada persona

pago = por_persona(total, amigos)

print("Cada persona paga:", pago)
```

```python
# Resultado

print(f"Cuenta original: ${cuenta:,.0f}")
print(f"Total con propina: ${total:,.0f}")
print(f"Cada persona paga: ${pago:,.0f}")
```

```python
# Creación función descuento

def descuento(total, porcentaje):

    return total - (total * porcentaje / 100)
```

```python
# Resultado

nuevo_total = descuento(total, 10)

print(f"Total con descuento: ${nuevo_total:,.0f}")
```

```python
# Ejercicio 10: La caja de la tienda

precios = [120000, 180000, 250000, 95000]
```

```python
# Crear acumulador

total = 0
```

```python
# Suma con FOR

for precio in precios:
    total += precio
```

```python
# Verificar el total

print("Total compra:", total)
```

```python
# Calcular descuento con IF

if total > 500000:
    descuento = total * 0.10
else:
    descuento = 0
```

```python
# Verificar el descuento

print("Descuento:", descuento)
```

```python
# Cacular valor final

a_pagar = total - descuento
```

```python
# Verificar valor final

print("A pagar:", a_pagar)
```

```python
# Imprimir el ticket con f-strings

print(f"Total: ${total:,.0f}")
print(f"Descuento: ${descuento:,.0f}")
print(f"A pagar: ${a_pagar:,.0f}")
```
